```python
# Se decide crear una tabla maestra de geolocalización que contenga: la media para los valores de latitud y longitud, y la moda para la ciudad y el país.

geolocation_master = (geolocation_clean.groupby("geolocation_zip_code_prefix").agg({
    "geolocation_lat": "mean",
    "geolocation_lng": "mean",
    "geolocation_city" : lambda x: x.mode().iloc[0],
    "geolocation_state": lambda x: x.mode().iloc[0]
}). reset_index())
```

```python
geolocation_master.shape
```

La tabla geolocation contenía más de un millón de registros, aunque únicamente existían 19.015 códigos postales distintos.
Con el objetivo de facilitar las uniones con las tablas de clientes y vendedores, se creó una tabla maestra denominada geolocation_master, compuesta por un único registro por código postal. Esta transformación permitió reducir la tabla de más de un millón de registros a 19.015 observaciones sin perder la capacidad de representar geográficamente cada código postal.

FINALIZADO EL DATAFRAME DE VENTAS

Tras evaluar la posibilidad de incorporar también las tablas order_payments y order_reviews, se decidió mantenerlas como tablas independientes debido a que su nivel de granularidad es diferente. Mientras que order_items almacena una fila por producto vendido, order_reviews registra una valoración por pedido y order_payments puede contener varios registros para un mismo pedido cuando este se ha abonado mediante varios métodos de pago.
Integrar estas tablas directamente en el dataset de ventas habría provocado la repetición de información y la alteración de la granularidad de los datos, generando resultados incorrectos en agregaciones y análisis posteriores. Por este motivo, ambas tablas se conservaron independientes y se utilizarán mediante uniones puntuales únicamente cuando una pregunta de análisis requiera combinar información de distintos niveles del modelo.
